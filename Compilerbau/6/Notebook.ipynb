# Post mortem

Jede Person beschreibt in der ILIAS-Abgabe individuell(!) die Bearbeitung des jeweiligen Aufgabenblattes
zurückblickend mit ca. 200 bis 400 Wörtern. Gehen Sie dabei aussagekräftig und nachvollziehbar auf folgende Punkte ein: 
 (a) Zusammenfassung: Was wurde gemacht? 
 (b) Implementierungsdetails: Kurze Beschreibung besonders interessanter Aspekte der Umsetzung. 
 (c) Was war der schwierigste Teil bei der Bearbeitung? Wie haben Sie dieses Problem gelöst? 
 (d) Was haben Sie gelernt oder (besser) verstanden? 
 (e) Team: Mit wem haben Sie zusammengearbeitet? 
 (f) Link zum Repo mit der Lösung 



 
 * a: Ich habe ein C# Projekt aufgesetzt und die gegebene Antlr Grammatik in das Projekt kopiert. Stellenweise habe ich die Grammatik um labels erweitert, um das Parsen des Antlr Baums zu vereinfachen. Außerdem habe ich mir test Dateien geschrieben, um den Parser zu testen. Diese habe ich im laufe der Entwicklung erweitert. Als nächstes habe ich die Ast Struktur und Parser implementiert. Dabei habe ich mich an die Referenz aus dem Aufgabenblatt gehalten. Zuletzt habe ich den Generator für die Symboltabelle implementiert.
 * b: 
   + Ein Scope referenziert "rückwärts" auf den Eltern Scope. Das ist ein Konzept, welches ich schon mehrfach im KI Praktikum angewendet habe. 
   + Ich habe einen Visitor benutzt um den Antlr Baum in den Ast zu Parsen. Allerdings hatte ich in einem vorherigen Praktikum festgestellt, dass das Visitor interface eher umständlich ist, da alle Methoden den gleichen Rückgabewert haben. Dadurch war es ständig nötig den Rückgabewert zu casten, ohne das durch das interface ein wirklicher mehrwert entstanden ist. Daher habe ich die Funktionen dieses mal direkt implementiert.
 * c: Meine derzeitige Lösung spannt derzeit 3 Bäume auf. Den Ast und die SymbolTabelle die "parallel" verlaufen und die Scopes die rückwärts gehen. Ich bin mir noch nicht sicher, wie ich das am elegantesten lösen. Evtl. würde ich für einen tatsächlichen Interpreter einen neuen Ast / Symbol Baum erzeugen, der den Urprünglichen bereits weg wirft und den Baum auf einfachere (ausführbare) Komponenten reduziert.
 * d: Ich finde es gut, dass meine Implementation des Visitors diesmal deutlich sauberer ist.
