# PDA - PushDown-Automata

Der DFA wird um eine Stack erweitert.

 (PDA) P = (Q, Σ, Γ, δ, q0 , ⊥, F )

 * Q: Endliche Menge an Zuständen
 * Σ: Endliche Menge an Symbolen
 * Γ: Endliches Kelleralphabet
 * δ: Übergangsfunktion: δ: Q x Σ u {ϵ} x Γ -> 2^(QxΓ*)
 * q0: Startzustand
 * ⊥: Anfänglicher Stack (Keller) Inhalt. ⊥ ist ein leerer Stack
 * F: Menge von endzuständen

## Übergangsfunktion
Γ -> Pops the top element from the stack
gibt einen Zustand und beliebig viele Stack-elemente zurück  

Ein PDA ist Nicht-deterministisch
