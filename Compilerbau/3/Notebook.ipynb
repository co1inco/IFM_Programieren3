# PDA - PushDown-Automata

Der DFA wird um eine Stack erweitert.

 (PDA) P = (Q, Σ, Γ, δ, q0 , ⊥, F )

 * Q: Endliche Menge an Zuständen
 * Σ: Endliche Menge an Symbolen
 * Γ: Endliches Kelleralphabet
 * δ: Übergangsfunktion: δ: Q x Σ u {ϵ} x Γ -> 2^(QxΓ*)
 * q0: Startzustand
 * ⊥: Anfänglicher Stack (Keller) Inhalt. ⊥ ist ein leerer Stack
 * F: Menge von endzuständen

## Übergangsfunktion
Γ -> Pops the top element from the stack
gibt einen Zustand und beliebig viele Stack-elemente zurück  

Ein PDA ist Nicht-deterministisch

# Aufgabe 3

Context freie Grammatik


G = (N, T , P, S)

X → Y mit X ∈ N, Y ∈ (N ∪ T ) ∗ 

X ist ein element aus den Nicht-Terminalen
Y ist eine element ais den Nicht-Terminalen und Terminalen


N, T, S wie in formaler Grammatik
N: endliche Menge an Nicht-terminalen
T: endliche Menge an Terminalen ( N ∩ T = ∅ )
S ∈ N: Startsymbol



```
    G = ( { Statement , Condition , … } , { "if" , "else" , … } , P , Statement )
mit
    P = { 
        Statement → "if" Condition Statement | "if" Condition Statement "else" Statement
        Condition → …   
    } 
```

```
S -> "if" CS | "if" CS "else S | ...
C -> ...
```

~~Die Sprache ist mehrdeutig, da das "if" an mehreren stellen verwendet wird.~~

```"if" C "if" C S "else" S```
Das müsste der Grammatik entsprechen. Es ist aber nicht eindeutig, ob das "else" zu dem ersten oder zweiten "if" gehört

```
S -> "if" C "{" S "}" | "if" C "{" S "}" "else" "{" S "}"
```
