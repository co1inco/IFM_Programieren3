```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import sklearn
```

# Augabe 1

```python
COLUMN_TEMP_2M = "temperature_2m (°C)"

df = pd.read_csv("herford_weather.csv", parse_dates=['time']).set_index('time')
display(df.index.dtype)
df[:5]
```

```python
df[COLUMN_TEMP_2M].plot.line(figsize=(15,5))
```

# Aufgabe 2

```python
df_2022 = df['2022-01-01':'2022-12-31']
df_2022[:5]
```

```python
df_2022[COLUMN_TEMP_2M].plot.line(figsize=(15,5))
```

```python
df_2022_avg = df_2022.resample('d').mean()
df_2022_avg[COLUMN_TEMP_2M].plot.line(figsize=(15,5))
```

df[...] does not support a single value for selecting the index. To do that, use df.loc

```python
df.loc['2022-08-01'][COLUMN_TEMP_2M].plot.line(figsize=(15,5))
```

```python
df_broken = df['2022-08-01':'2022-08-01'] #.drop(pd.to_datetime('2022-08-01 18:00'))
df_broken.loc['2022-08-01 18:00', COLUMN_TEMP_2M] = np.nan # apparently setting a value to nan this way, changes the datatype for some FUCKING reason 
# updated. It magically fixed itself ...
# df_broken.replace(to_replace=pd.to_datetime('2022-08-01 18:00'), value=np.nan, inplace=True)
# df_broken.loc[df_broken[COLUMN_TEMP_2M] == pd.to_datetime('2022-08-01 18:00'), COLUMN_TEMP_2M] = np.nan
display(df_broken[COLUMN_TEMP_2M].dtypes)
display(df_broken[16:19])

df_broken[COLUMN_TEMP_2M].plot.line(figsize=(15,5))
```

```python
df_fixed = df_broken.copy()
mean = df_fixed.mean()

df_fixed[COLUMN_TEMP_2M] = df_fixed[COLUMN_TEMP_2M].fillna(mean[COLUMN_TEMP_2M])
df_fixed[COLUMN_TEMP_2M].plot.line(figsize=(15,5))
```

```python
df_fixed = df_broken.copy()

df_fixed = df_fixed.ffill()
df_fixed[COLUMN_TEMP_2M].plot.line(figsize=(15,5))
```
