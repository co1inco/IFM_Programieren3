# Aufgabe 2

## Population & Density

```python
states = pd.read_csv("bundeslaender.txt", sep=" ")
# states = states[["land", "area", "female", "male"]]

states["population"] = states["male"] + states["female"]

states["density"] = ((states["population"] * 1000) / states["area"]).round()

states.to_csv("density.txt", sep=" ")

states
```

## Mehr Frauen als Männer

```python
states_more_females = states.loc[states["female"] > states["male"]]

print("Anzahl der Bundesländer mit größerer weiblicher Bevölkerung " + str(states_more_females.shape[0]))
states_more_females
```

## > 1000 Einwohner pro km2

```python
states.loc[((states["population"] * 1000) / states["area"]) > 1000]
```

## Reguläre Wochentage

```python
index = pd.date_range("2025-12-24", "2026-1-6", freq="B")

print("Es gibt " + str(len(index)) + " normale Wochentage/reguläre Arbeitstage in den Weihnachtsferien 2025/26.")
index
```

## Sonntage in 2026

```python
index2 = pd.date_range("2025-12-24", "2027-01-06")
index2 = index2[(index2.day == 1) & (index2.weekday == 6)]

print("Vom 24.12.2025 bis zum 6.01.2027 gibt es " + str(len(index2)) + " Sonntage, die auf den 1. des Monats fallen.")

index2
```
